```python
import numpy as np
import pandas as pd
from pathlib import Path
import os
from scipy.stats import chi2_contingency
```

```python
#Load and merge data
BASE_DIR = Path("../../data")
RAW_DIR = BASE_DIR / "raw" / "widsdatathon2025"
FEATURE_ENGINEERED_DIR = BASE_DIR / "feature engineered"

df_feateng_quant = pd.read_csv(FEATURE_ENGINEERED_DIR/ "wids_feature_engineered_quantitative.csv")
df_feateng_cat = pd.read_csv(FEATURE_ENGINEERED_DIR/ "wids_feature_engineered_categorical.csv")
df_feateng_fcm= pd.read_csv(FEATURE_ENGINEERED_DIR/ "wids_feature_engineered_fcm.csv")
df_labels = pd.read_excel(RAW_DIR / "TRAIN/TRAINING_SOLUTIONS.xlsx")

# combined_df_feat_quant = pd.merge(df_feateng_quant, df_labels, on='participant_id')
# combined_df_feat_quant.head()
df_feateng_quant.head()
```

```python
# List of quant columns (excluding participant_id & ADHD_Outcome)
quant_columns = [col for col in df_feateng_quant .columns if col not in ['participant_id', 'ADHD_Outcome']]
print("Chi-Square test of demographic variables and adhd outcome")
# Loop through categorical variables and run Chi-Square test
significant_quant = {} #initialize dictionary for p values that are < 0.05
for col in quant_columns:
    contingency_table = pd.crosstab([col], ['ADHD_Outcome'])
    chi2, p, _, _ = chi2_contingency(contingency_table)
    print(f"{col}: Chi-Square = {chi2:.3f}, p-value = {p:.5f}")
    if p < 0.05: #if p val < 0.05, add to dict
        significant_quant[col] = p
```

```python
print("-----------------------")
print("Significant Quant Preprocessed Features with ADHD Outcome (p < 0.05):")
for col, p in significant_quant.items(): #print dict w significant p vals
    print(f"{col}: p-value = {p:.5f}")
```

```python
#Load and merge categorical data
BASE_DIR = Path("../../data")
RAW_DIR = BASE_DIR / "raw" / "widsdatathon2025"

train_cat = pd.read_excel(RAW_DIR / "TRAIN/TRAIN_CATEGORICAL_METADATA.xlsx")
train_quant = pd.read_excel(RAW_DIR / "TRAIN/TRAIN_QUANTITATIVE_METADATA.xlsx")
train_solutions = pd.read_excel(RAW_DIR / "TRAIN/TRAINING_SOLUTIONS.xlsx")

combined_df_cat= pd.merge(train_cat, train_solutions, on='participant_id')
combined_df_quant = pd.merge(train_quant, train_solutions, on='participant_id')
```

```python
# List of numerical columns (excluding participant_id & ADHD_Outcome)
num_columns = [col for col in combined_df_quant.columns if col not in ['participant_id', 'ADHD_Outcome']]
print("Chi-Square test of numerical variables and adhd outcome")
# Loop through numerica; variables and run Chi-Square test
significant_num = {}
for col in num_columns:
    contingency_table = pd.crosstab(combined_df_quant[col], combined_df_quant['ADHD_Outcome'])
    chi2, p, _, _ = chi2_contingency(contingency_table)
    print(f"{col}: Chi-Square = {chi2:.3f}, p-value = {p:.5f}")
    if p < 0.05:
        significant_num[col] = p
```

```python
print("-----------------------")
print("Significant Numerical Features with ADHD Outcome (p < 0.05):")
for col, p in significant_num.items():
    print(f"{col}: p-value = {p:.5f}")
```

```python
# Drop statistically insignificant numerical features
insignificant_columns = [col for col in num_columns if col not in significant_num]

train_quant = train_quant.drop(columns=insignificant_columns)

print("\nDropped the following insignificant numerical columns:")
print(insignificant_columns)
```

```python
# List of numerical columns (excluding participant_id & ADHD_Outcome)
cat_columns = [col for col in combined_df_cat.columns if col not in ['participant_id', 'ADHD_Outcome']]
print("Chi-Square test of numerical variables and adhd outcome")
# Loop through numerica; variables and run Chi-Square test
significant_cat = {}
for col in cat_columns:
    contingency_table = pd.crosstab(combined_df_cat[col], combined_df_cat['ADHD_Outcome'])
    chi2, p, _, _ = chi2_contingency(contingency_table)
    print(f"{col}: Chi-Square = {chi2:.3f}, p-value = {p:.5f}")
    if p < 0.05:
        significant_cat[col] = p
```

```python
print("-----------------------")
print("Significant Cat Features with ADHD Outcome (p < 0.05):")
for col, p in significant_cat.items():
    print(f"{col}: p-value = {p:.5f}")
```

```python
# Drop statistically insignificant categorical features
insignificant_columns = [col for col in cat_columns if col not in significant_cat]

train_cat = train_cat.drop(columns=insignificant_columns)

print("\nDropped the following insignificant numerical columns:")
print(insignificant_columns)
```

```python
# Define the output file path
PVAL_DIR = BASE_DIR / "dropped p values"
output_path_cat = PVAL_DIR / "cat_dropped_pvals.csv"
output_path_quant = PVAL_DIR / "quant_dropped_pvals.csv"
# Save the filtered DataFrame to CSV
train_cat.to_csv(output_path_cat, index=False)
train_quant.to_csv(output_path_quant, index=False)
```
